```python
# Praktikum 1: Perulangan for dengan range ( )
# Tujuan: Mahasiswa dapat menggunakan fungsi range() dengan berbagai parameter untuk mengontrol urutan angka yang dihasilkan.

# Langkah 2 — Ketik dan jalankan: Bilangan Ganjil 1–9

# === PRAKTIKUM 1A: Bilangan Ganjil 1–9 ===
print("--- Bilangan Ganjil 1 hingga 9 ---")
for angka in range(1, 10, 2):   # start=1, stop=10, step=2
    print(angka)
```

```python
# Praktikum 2: Perulangan while dan Nilai Penghenti
# Tujuan: Mahasiswa memahami cara kerja while dan menggunakan nilai Nilai Penghenti untuk mengontrol penghentian perulangan.

# Langkah 1 — Perulangan while Dasar

# === PRAKTIKUM 2A: while Dasar ===
hitung = 1
while hitung <= 5:
    print(f"Iterasi ke-{hitung}")
    hitung += 1  # sangat penting: tanpa ini, loop tak akan berhenti
print("Perulangan selesai.")
```

```python
# Langkah 2 — Nilai Penghenti: Berhenti dengan Input 99

# === PRAKTIKUM 2B: Nilai Penghenti ===
kode_berhenti = 99
input_pengguna = 0

while input_pengguna != kode_berhenti:
    try:
        input_pengguna = int(input("Masukkan angka (99 untuk berhenti): "))
        if input_pengguna != kode_berhenti:
            print(f"  → Anda memasukkan: {input_pengguna}")
    except ValueError:
        print("  Input tidak valid. Harap masukkan angka.")

print("Perulangan telah dihentikan oleh Nilai Penghenti (99).")
```

```python
# Langkah 3 — Modifikasi: Berhenti dengan NIM

# === PRAKTIKUM 2C: Nilai Penghenti dengan NIM ===
NIM = 4232501048   # ganti dengan NIM Anda
input_pengguna = 0

while input_pengguna != NIM:
    try:
        input_pengguna = int(input("Masukkan NIM untuk berhenti: "))
        if input_pengguna != NIM:
            print("  NIM belum cocok, coba lagi.")
    except ValueError:
        print("  Input tidak valid.")

print(f"NIM {NIM} cocok. Program berhenti.")
```

```python
# Langkah 4 — Batasan Percobaan (Limitasi Iterasi)

# === PRAKTIKUM 2D: Batas Percobaan Password ===
percobaan_maksimal = 3
percobaan_saat_ini = 1
password_benar = "polibatam"

while percobaan_saat_ini <= percobaan_maksimal:
    masukan = input(f"Percobaan ke-{percobaan_saat_ini}: Masukkan password: ")
    if masukan == password_benar:
        print("Akses diterima!")
        break
    else:
        print("  Password salah.")
    percobaan_saat_ini += 1
else:
    print("Kesempatan habis. Akses ditolak.")
```

```python
# Praktikum 3: Iterasi pada List dan String
# Tujuan: Mahasiswa dapat menggunakan for untuk menelusuri elemen list dan karakter string, serta memanfaatkan enumerate() untuk mendapatkan indeks sekaligus nilai.

# Langkah 1 — Iterasi pada List: Menghitung Total Belanja

# === PRAKTIKUM 3A: Total Belanja ===
harga_barang = [15000, 25000, 10000, 5000, 8000,
                12000, 30000, 7500, 22000, 18000,
                9500, 4000]   # minimal 10 item
total_belanja = 0

for harga in harga_barang:
    total_belanja += harga

print(f"Jumlah item  : {len(harga_barang)}")
print(f"Total belanja: Rp {total_belanja:,.0f}")
```

```python
# Praktikum 4: Kontrol Perulangan: break dan continue
# Tujuan: Mahasiswa dapat menggunakan break untuk menghentikan perulangan dan continue untuk melewati iterasi tertentu.

# Langkah 1 — Menggunakan break untuk Menghentikan Loop

# === PRAKTIKUM 4A: break saat menemukan angka -1 ===
data_angka = [10, 20, 30, -1, 50, 60]

print("Memproses data:")
for angka in data_angka:
    if angka == -1:
        print("  → Angka -1 ditemukan! Proses dihentikan.")
        break
    print(f"  Memproses: {angka}")

print("Program selesai.")
```

```python
# Langkah 2 — Modifikasi: Berhenti saat Angka 0

# === PRAKTIKUM 4B: break saat menemukan angka 0 ===
data_sensor = [45.2, 67.8, 89.1, 0, 55.3, 72.4]

print("Membaca data sensor pembangkit:")
for nilai in data_sensor:
    if nilai == 0:
        print("  ⚠ Pembacaan sensor = 0! Kemungkinan sensor mati. Proses dihentikan.")
        break
    print(f"  Suhu terbaca: {nilai}°C")
```

```python
# Langkah 3 — Menggunakan continue untuk Melewati Iterasi

# === PRAKTIKUM 4C: continue untuk melewati angka negatif ===
data_daya = [120.5, -1, 135.2, 0.0, 98.7, -2, 145.0]

print("Data daya valid (MW):")
for daya in data_daya:
    if daya <= 0:
        print(f"  (Melewati data tidak valid: {daya})")
        continue
    print(f"  Daya: {daya} MW")
```

```python
# Langkah 4 — Kombinasi break dan continue

# === PRAKTIKUM 4D: Penambahan berkelanjutan sampai total > 20 ===
total = 0
bilangan = 1
batas = 20

while total <= batas:
    total += bilangan
    bilangan += 1

print(f"Total pertama yang melebihi {batas}: {total}")
print(f"Bilangan terakhir ditambahkan: {bilangan - 1}")
```

```python
# Praktikum 5: Perulangan Bersarang (Nested Loop)
# Tujuan: Mahasiswa dapat membuat dan memahami perulangan bersarang untuk mencetak pola dan memproses data multi-dimensi.

# Langkah 1 — Nested Loop Dasar: Pola Teks

# === PRAKTIKUM 5A: Mencetak "Python" 3 baris ===
jumlah_baris = 3

for baris in range(jumlah_baris):
    baris_teks = ""
    for _ in range(1):     # inner loop: 1 kata per baris
        baris_teks += "Python "
    print(baris_teks.strip())
```

```python
# === PRAKTIKUM 5A: Mencetak "Python" 3 baris ===
jumlah_baris = 3

for baris in range(jumlah_baris):
    baris_teks = ""
    for _ in range(3):     # inner loop: 1 kata per baris
        baris_teks += "Python "
    print(baris_teks.strip())
```

```python
# Langkah 2 — Pola Bintang dengan Nested Loop

# === PRAKTIKUM 5B: Pola Segitiga Bintang ===
print("Pola Segitiga:")
for baris in range(1, 6):           # baris 1 sampai 5
    for kolom in range(baris):      # kolom bertambah sesuai baris
        print("*", end=" ")
    print()  # pindah baris
```

```python
# === PRAKTIKUM 5B: Pola Segitiga Bintang ===
print("Pola Segitiga:")
for baris in range(1, 6):           # baris 1 sampai 5
    for kolom in range(5, baris, -1):      # kolom bertambah sesuai baris
        print("*", end=" ")
    print()  # pindah baris
```

```python
# Langkah 3 — Nested Loop untuk Data Sensor Multi-Hari

# === PRAKTIKUM 5C: Tabel Suhu Sensor (Minggu × Hari) ===
nama_unit = ["Unit 1", "Unit 2", "Unit 3"]
suhu_harian = [
    [85.2, 87.1, 86.5, 88.0, 84.9, 85.7, 86.2],   # Unit 1
    [90.1, 91.3, 89.7, 92.0, 90.5, 91.1, 90.8],   # Unit 2
    [78.4, 79.2, 77.8, 80.1, 79.5, 78.9, 79.7],   # Unit 3
]
hari = ["Sen", "Sel", "Rab", "Kam", "Jum", "Sab", "Min"]

print("Laporan Suhu Boiler (°C)")
print(f"{'Unit':<10}", end="")
for h in hari:
    print(f"{h:>6}", end="")
print(f"{'Rata2':>8}")
print("-" * 60)

for i, unit in enumerate(nama_unit):   # outer loop: per unit
    print(f"{unit:<10}", end="")
    total = 0
    for suhu in suhu_harian[i]:         # inner loop: per hari
        print(f"{suhu:>6.1f}", end="")
        total += suhu
    rata2 = total / len(suhu_harian[i])
    print(f"{rata2:>8.1f}")
```

```python
# Nama : Tiara Putri Az Zahra
# Nim : 4232501048
# Tanggal : 14, Oktober 2026
# Tujuan : Menghitung nilai faktorial bilangan 1-5 menggunakan percabangan if, elif dan else serta memvalidasi input.

# Soal 1 - Kalkulator Faktorial (1-5)

n = int(input("Masukkan bilangan bulat (1-5): "))

# Validasi input
if n < 1 or n > 5:
    print("Input tidak valid. Masukkan bilangan antara 1 sampai 5.")
elif n == 1:
    hasil = 1
    print("1! =", hasil)
elif n == 2:
    hasil = 2 * 1
    print("2! =", hasil)
elif n == 3:
    hasil = 3 * 2 * 1
    print("3! =", hasil)
elif n == 4:
    hasil = 4 * 3 * 2 * 1
    print("4! =", hasil)
else:
  hasil = 5 * 4 * 3 * 2 * 1
  print("5! =", hasil)
```

```python
# Nama : Tiara Putri Az Zahra
# Nim : 4232501048
# Tanggal : 14, Oktober 2026
# Tujuan : Menghitung estimasi daya panel dan menentukan mode operasi PLTS menggunakan percabangan if, elif dan else.

# Soal 3 - Rekomendasi mode operasi PLTS berdasarkan iradiasi dan kapasitas baterai

iradiasi = float(input("Masukkan iradiasi matahari (W/m²): "))
soc = float(input("Masukkan SoC baterai (%): "))

# Menghitung estimasi daya panel
daya_panel = iradiasi * 0.18 * 10

print(f"Estimasi daya panel: {daya_panel:.1f} W")

# Menentukan mode operasi
if iradiasi > 600 and soc < 80:
  mode = "CHARGING – prioritaskan pengisian baterai"
elif iradiasi > 600 and soc >= 80:
  mode = "GRID-EXPORT – ekspor kelebihan daya ke jaringan"
elif iradiasi > 200 and soc >= 20:
  mode = "NORMAL – operasi mandiri"
elif iradiasi <= 200 and soc >= 50:
  mode = "BATTERY-ONLY – andalkan baterai"
else:
  mode = "EMERGENCY – kurangi beban segera"

print(f"Iradiasi {iradiasi} W/m², SoC {soc}% → Mode: {mode}")
```

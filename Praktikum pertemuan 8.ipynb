**Aktivitas 6E - Salinan atau bukan?**

Tanpa menjalankan kode, tebak isi x, y, dan z setelah baris-baris ini, lalu periksa dengan blok di atas: x = [10, 20, 30], y = x, z = x[:], y.append(40), z.append(50), x.insert(0, 5). Mengapa z berbeda?

```python
# Nama : Tiara Putri Az Zahra
# NIM : 4232501048
# Tanggal : 17 November 2026

# Aktivitas 6E - Salinan atau bukan?

# Membuat list awal
x = [10, 20, 30]

# y menjadi nama lain dari list x (bukan salinan)
y = x

# z menjadi salinan baru dari list x menggunakan slicing
z = x[:]

# Menambahkan elemen 40 melalui y
y.append(40)

# Menambahkan elemen 50 melalui z
z.append(50)

# Menambahkan elemen 5 di awal list x
x.insert(0, 5)

# Menampilkan hasil akhir
print("Isi x:", x)
print("Isi y:", y)
print("Isi z:", z)
```

**Jawaban pertanyaan Aktivitas 6E**

Mengapa z berbeda?

Karena z = x[:] membuat salinan baru dari list x. Artinya, z memiliki list tersendiri sehingga perubahan pada x atau y tidak memengaruhi z.

Sementara itu, y = x bukan salinan, melainkan hanya membuat nama lain untuk list yang sama. Oleh karena itu, perubahan pada x juga akan terlihat pada y, begitu pula sebaliknya.

**Aktivitas 6F - Tebak dulu, baru jalankan**

Lanjutkan tabel secara manual untuk dua elemen terakhir (75 dan 70). Berapa total akhir dan rata-ratanya? Berapa skor yang bernilai 75 atau lebih? Lalu jalankan blok berikut.

```python
# Nama : Tiara Putri Az Zahra
# NIM : 4232501048
# Tanggal : 17 November 2026

# Aktivitas 6F - Tebak dulu, baru jalankan

# Membuat list nilai
nilai = [80, 65, 90, 75, 70]

# Menginisialisasi variabel akumulator
total = 0
jumlah_lulus = 0

# Melakukan perulangan untuk menghitung total dan jumlah skor >= 75
for skor in nilai:
    total += skor

    if skor >= 75:
        jumlah_lulus += 1

# Menghitung rata-rata
rata_rata = total / len(nilai)

# Menampilkan hasil
print("Daftar nilai:", nilai)
print("Total nilai:", total)
print("Rata-rata nilai:", rata_rata)
print("Jumlah skor 75 atau lebih:", jumlah_lulus)
```

**Aktivitas 6G - Ganti ukuran jendela**

Ubah blok di atas menjadi rata-rata bergerak 5 hari (mulai perulangan di indeks 4, slice suhu[i - 4 : i + 1], bagi dengan 5). Berapa rata-rata yang Anda dapat dari 7 nilai? Mengapa?

```python
# Nama : Tiara Putri Az Zahra
# NIM : 4232501048
# Tanggal : 17 November 2026

# Aktivitas 6G - Ganti ukuran jendela

suhu = [28, 31, 30, 33, 29, 27, 32]

# Perubahan harian: mulai dari indeks 1 karena kita melihat i - 1
for i in range(1, len(suhu)):
    perubahan = suhu[i] - suhu[i - 1]
    print(f"hari {i + 1}: {suhu[i]} °C, perubahan {perubahan:+d}")

# Rata-rata bergerak 5 hari: jendela = 5 nilai yang berakhir di indeks i
rata_5 = []
for i in range(4, len(suhu)):
    jendela = suhu[i - 4 : i + 1]
    rata_5.append(round(sum(jendela) / 5, 1))

print("Rata-rata 5 hari:", rata_5)
print(len(suhu), "nilai ->", len(rata_5), "rata-rata")
```

Energi hari ke-17 turun lebih dari 40%. Dalam kelompok, tuliskan tiga alasan operasional yang mungkin menjelaskannya dan data apa yang akan Anda periksa untuk memastikan masing-masing. (Petunjuk: lihat jam_operasi[16] di bagian data 30 hari, dan blok hari kerja di bawah.)

**Tiga kemungkinan penyebab:**

1. Jam operasi berkurang karena perawatan atau penghentian sementara.
   Data yang diperiksa: run_hours (hari ke-17 hanya 18 jam).

2. Terjadi gangguan atau kerusakan pada pembangkit.
   Data yang diperiksa: log gangguan, alarm, atau laporan maintenance.

3. Permintaan beban listrik menurun sehingga produksi energi ikut berkurang.

```python
# Nama : Tiara Putri Az Zahra
# NIM : 4232501048
# Tanggal : 17 November 2026

# Aktivitas 6H - Jelaskan penurunannya

energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]

days = list(range(1, 31))
run_hours = [24] * 30
run_hours[16] = 18  # Hari ke-17 hanya beroperasi 18 jam

DAY_NAMES = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]

idx = 16  # Hari ke-17

persen_ubah = (energy[idx] - energy[idx - 1]) / energy[idx - 1] * 100
rata_rata_sebelumnya = sum(energy[idx - 5 : idx]) / 5

print(f"Hari ke-{days[idx]} ({DAY_NAMES[idx % 7]}):")
print(f"  Energi hari ke-16 : {energy[idx - 1]} MWh ({run_hours[idx - 1]} jam)")
print(f"  Energi hari ke-17 : {energy[idx]} MWh ({run_hours[idx]} jam)")
print(f"  Perubahan harian  : {persen_ubah:+.1f}%")
print(f"  Rata-rata 5 hari sebelumnya : {rata_rata_sebelumnya:.1f} MWh")
```

**Aktivitas 6I - Array Anda sendiri**

Buat list suhu berisi 7 suhu harian (bebas). Lalu: (1) tampilkan 3 hari pertama dan 2 hari terakhir dengan slicing; (2) tambahkan satu hari dengan append, lalu keluarkan lagi dengan pop; (3) tampilkan versi urutnya dengan sorted tanpa mengubah suhu; (4) hitung rata-rata dan cari suhu tertinggi beserta hari ke berapa; (5) buat list baru berisi suhu di atas rata-rata.

```python
# Nama : Tiara Putri Az Zahra
# NIM : 4232501048
# Tanggal : 17 November 2026

# Aktivitas 6I - Array Anda sendiri

# List suhu selama 7 hari
suhu = [28, 30, 29, 32, 31, 27, 33]

# 1. Menampilkan 3 hari pertama dan 2 hari terakhir
print("3 hari pertama:", suhu[:3])
print("2 hari terakhir:", suhu[-2:])

# 2. Menambahkan satu suhu dengan append, lalu mengeluarkannya dengan pop
suhu.append(34)
print("Setelah append:", suhu)

suhu.pop()
print("Setelah pop:", suhu)

# 3. Menampilkan suhu dalam urutan terkecil ke terbesar
print("Suhu terurut:", sorted(suhu))

# 4. Menghitung rata-rata dan mencari suhu tertinggi
rata_rata = sum(suhu) / len(suhu)
tertinggi = max(suhu)
hari = suhu.index(tertinggi) + 1

print("Rata-rata suhu:", rata_rata)
print("Suhu tertinggi:", tertinggi)
print("Suhu tertinggi terdapat pada hari ke-", hari)

# 5. Membuat list suhu di atas rata-rata
di_atas_rata = []

for nilai in suhu:
    if nilai > rata_rata:
        di_atas_rata.append(nilai)

print("Suhu di atas rata-rata:", di_atas_rata)
```

**Pemanasan - Lima latihan kecil (data kecil)**

Pakai skor = [80, 65, 90, 75, 70, 85]. Tulis kode yang (1) mencetak skor pertama dan skor terakhir (pakai indeks negatif untuk yang terakhir), (2) mencetak rata-rata, (3) menghitung berapa skor di atas rata-rata, (4) mencetak posisi skor terendah sebagai nomor mahasiswa (mahasiswa pertama = 1), dan (5) membuat list baru dengan setiap skor ditambah 5.

```python
# Nama : Tiara Putri Az Zahra
# NIM : 4232501048
# Tanggal : 17 November 2026

# Pemanasan - Lima latihan kecil (data kecil)

skor = [80, 65, 90, 75, 70, 85]

# 1. Mencetak skor pertama dan terakhir
print("1. Skor pertama:", skor[0])
print("   Skor terakhir:", skor[-1])

# 2. Menghitung rata-rata
rata_rata = sum(skor) / len(skor)
print("2. Rata-rata skor:", rata_rata)

# 3. Menghitung jumlah skor di atas rata-rata
jumlah = 0
for nilai in skor:
    if nilai > rata_rata:
        jumlah += 1
print("3. Jumlah skor di atas rata-rata:", jumlah)

# 4. Mencari posisi skor terendah
terendah = min(skor)
nomor = skor.index(terendah) + 1
print("4. Skor terendah:", terendah)
print("   Nomor mahasiswa:", nomor)

# 5. Membuat list baru dengan setiap skor ditambah 5
skor_baru = []
for nilai in skor:
    skor_baru.append(nilai + 5)
print("5. Skor setelah ditambah 5:", skor_baru)
```

**Latihan praktikum - Laporan produksi bulanan (portofolio)**

Dengan list energi 30 hari (salin dari blok kode di atas), tulis program yang mencetak: total dan rata-rata energi; tiga hari tertinggi dan tiga hari terendah (nomor hari dan nilainya); jumlah hari di bawah target yang Anda masukkan; setiap hari yang rata-rata bergerak 7 harinya turun di bawah 97% rata-rata bulanan; serta total energi per minggu (minggu 1–4, ditambah sisa hari).

```python
# Nama : Tiara Putri Az Zahra
# NIM : 4232501048
# Tanggal : 17 November 2026

# Latihan praktikum - Laporan produksi bulanan (portofolio)

energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062
]

# 1. Total dan rata-rata energi
total = sum(energi)
rata_rata = total / len(energi)

print("TOTAL ENERGI:", total, "MWh")
print("RATA-RATA ENERGI:", round(rata_rata, 2), "MWh")

# 2. Tiga hari tertinggi dan terendah
data_hari = []

for i in range(len(energi)):
    data_hari.append((energi[i], i + 1))

tertinggi = sorted(data_hari, reverse=True)
terendah = sorted(data_hari)

print("\nTIGA HARI TERTINGGI:")
for nilai, hari in tertinggi[:3]:
    print("Hari", hari, ":", nilai, "MWh")

print("\nTIGA HARI TERENDAH:")
for nilai, hari in terendah[:3]:
    print("Hari", hari, ":", nilai, "MWh")

# 3. Menghitung jumlah hari di bawah target
target = float(input("\nMasukkan target energi: "))

jumlah = 0
for nilai in energi:
    if nilai < target:
        jumlah += 1

print("Jumlah hari di bawah target:", jumlah)

# 4. Rata-rata bergerak 7 hari
batas = 0.97 * rata_rata

print("\nHARI DENGAN RATA-RATA BERGERAK DI BAWAH 97% RATA-RATA BULANAN:")

for i in range(6, len(energi)):
    rata_7 = sum(energi[i-6:i+1]) / 7

    if rata_7 < batas:
        print("Hari", i + 1, ":", round(rata_7, 2), "MWh")

# 5. Total energi per minggu
print("\nTOTAL ENERGI PER MINGGU:")

for i in range(0, len(energi), 7):
    minggu = energi[i:i+7]
    nomor_minggu = (i // 7) + 1

    print("Minggu", nomor_minggu, ":", sum(minggu), "MWh")
```

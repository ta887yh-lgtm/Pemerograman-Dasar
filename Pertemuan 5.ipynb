**Soal 1 : Belanja Buku**

Budi membeli 7 buku dengan harga Rp 18.500 per buku. Toko memberi diskon 10% dari total belanja. Budi membayar dengan uang Rp 150.000.

Tentukan:

Total belanja, besar diskon, dan jumlah yang harus dibayar.
Uang kembalian.
Jika jumlah yang dibayar dipatungi 4 orang secara adil dalam bilangan bulat, berapa iuran per orang dan berapa rupiah sisanya?
Apakah uang Budi cukup? (hasil True/False)

```python
# Nama : Tiara Putri Az Zahra
# Nim : 4232501048
# Tanggal : 07, Oktober 2026
# Soal 1 : Belanja Buku

jumlah_buku = int(input("Masukkan jumlah buku: "))
harga_per_buku = int(input("Masukkan harga per buku: "))
diskon_persen = int(input("Masukkan diskon (%): "))
uang_bayar = int(input("Masukkan uang yang dibayar: "))

# Perhitungan
total = jumlah_buku * harga_per_buku
diskon = total * diskon_persen // 100
bayar = total - diskon
kembalian = uang_bayar - bayar

# Dibagi rata ke 4 orang
per_orang = bayar // 4
sisa = bayar % 4

# Cek uang cukup
uang_cukup = uang_bayar >= bayar

print("Total :", total)
print("Diskon :", diskon)
print("Bayar :", bayar)
print("Kembalian :", kembalian)
print("Per orang :", per_orang)
print("Sisa :", sisa)
print("Uang cukup? :", uang_cukup)
```

**Soal 2 : PLTS Atap Rumah**

Sebuah rumah memasang 12 panel surya berdaya 450 Wp per panel. Rata-rata penyinaran efektif 5 jam/hari dengan efisiensi sistem 80% (0,8). Kebutuhan listrik rumah 25 kWh/hari dan tarif listrik Rp 1.444,70/kWh.

Tentukan:

Energi yang dihasilkan per hari (kWh).
Energi per bulan (30 hari) dan nilai rupiah listrik yang dihasilkan.
Kekurangan energi per hari terhadap kebutuhan.
Apakah PLTS sudah mencukupi kebutuhan rumah? (True/False)

```python
# Nama : Tiara Putri Az Zahra
# Nim : 4232501048
# Tanggal : 07, Oktober 2026
# 2. PLTS Atap Rumah

jumlah_panel = int(input("Masukkan jumlah panel surya: "))
daya_panel = int(input("Masukkan daya per panel (Wp): "))
jam_efektif = float(input("Masukkan lama penyinaran efektif (jam/hari): "))
efisiensi = float(input("Masukkan efisiensi sistem: "))
kebutuhan = float(input("Masukkan kebutuhan listrik rumah (kWh/hari): "))
tarif = float(input("Masukkan tarif listrik (Rp/kWh): "))

# Perhitungan
energi_harian = (jumlah_panel * daya_panel * jam_efektif * efisiensi) / 1000
energi_bulanan = energi_harian * 30
nilai_listrik = energi_bulanan * tarif
kekurangan = kebutuhan - energi_harian
mencukupi = energi_harian >= kebutuhan

print("Energi harian (kWh) :", round(energi_harian, 1))
print("Energi bulanan (kWh) :", round(energi_bulanan, 1))
print("Nilai listrik (Rp) :", round(nilai_listrik, 1))
print("Kekurangan (kWh) :", round(kekurangan, 1))
print("Mencukupi? :", mencukupi)
```

**Soal 3 : Generator Diesel dan Rugi Kabel**

Sebuah generator diesel 200 kW beroperasi 8 jam/hari. Konsumsi bahan bakar 0,25 liter per kWh, harga solar Rp 13.500/liter, dan kapasitas tangki 1.000 liter. Kabel penghantar dialiri arus 20 A dengan hambatan 0,5 Ω.

Tentukan:

Energi yang dibangkitkan per hari (kWh).
Kebutuhan solar per hari dan biayanya.
Berapa hari penuh tangki cukup, dan berapa liter solar yang tersisa?
Rugi daya pada kabel (P = I² × R).
Apakah rugi kabel tergolong tinggi, yaitu lebih dari 150 W dan arus minimal 20 A? (True/False)

```python
# Nama : Tiara Putri Az Zahra
# Nim : 4232501048
# Tanggal : 07, Oktober 2026
# 3. Generator Diesel dan Rugi Kabel

daya_generator = int(input("Masukkan daya generator (kW): "))
lama_operasi = int(input("Masukkan lama operasi (jam/hari): "))
konsumsi_solar = float(input("Masukkan konsumsi solar (liter/kWh): "))
harga_solar = int(input("Masukkan harga solar (Rp/liter): "))
kapasitas_tangki = int(input("Masukkan kapasitas tangki (liter): "))
arus = int(input("Masukkan arus (A): "))
hambatan = float(input("Masukkan hambatan kabel (ohm): "))

energi = daya_generator * lama_operasi
solar = energi * konsumsi_solar
biaya = solar * harga_solar

tangki_cukup = kapasitas_tangki // solar
sisa_solar = kapasitas_tangki % solar

rugi_kabel = arus ** 2 * hambatan
rugi_tinggi = rugi_kabel > 150 and arus >= 20

print("Energi (kWh) :", energi)
print("Solar (liter) :", solar)
print("Biaya (Rp) :", biaya)
print("Tangki cukup :", tangki_cukup, "hari")
print("Sisa solar (L) :", sisa_solar)
print("Rugi kabel (W) :", rugi_kabel)
print("Rugi tinggi? :", rugi_tinggi)
```

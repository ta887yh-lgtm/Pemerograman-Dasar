**1. Urutan Prioritas Operator (Operator Precedence)**

Dalam bahasa pemrograman Python, penggunaan beberapa operator dalam satu ekspresi mengikuti aturan prioritas operator (operator precedence). Aturan ini menentukan operasi yang harus dikerjakan terlebih dahulu. Pada operator aritmatika, perkalian, pembagian, pembagian bulat, dan modulus memiliki prioritas lebih tinggi dibandingkan penjumlahan dan pengurangan.

Sehingga pada ekspresi 10 + 5 * 2 operasi 5 * 2 yang akan dikerjakan terlebih dahulu, karena perkalian memiliki prioritas yang lebih tinggi. Hasilnya adalah 10, kemudian hasil tersebut ditambahkan dengan angka 10, sehingga diperoleh 10 + 10 = 20. Apabila penjumlahan yang ingin dilakukan terlebih dahulu, maka digunakan tanda kurung seperti (10 + 5) * 2 sehingga menghasilkan nilai 30.

Sumber: Python Software Foundation, Python 3 Documentation - Operator precedence.



**2. Operator Relasional dan Operator Logika**

Operator relasional (>, <, ==, >=, <=, dan !=), digunakan untuk membandingkan dua nilai. Hasil dari perbandingan tersebut berupa nilai True atau False. Sedangkan operator logika (and, or, not) digunakakan untuk menggabungkan atau membalik suatu kondisi sehingga menghasilkan nilai True atau False.

Contoh pada sistem penerimaan beasiswa berdasarkan nilai. misalnya, seorang mahasiswa dinyatakan lulus jika nilai ujian minimal 75 dan kehadirannya minimal 80%.

nilai = 80

kehadiran = 85

lulus = nilai >= 75 and kehadiran >= 80

print (lulus)

maka, nilai nya adalah True

**3. Contoh Kasus Nyata Tentang Bug atau Kesalahan Perhitungan Akibat Salah Memahami Operator Precedence**

Salah satu kasus nyata di forum Stack Overflow adalah kesalahan menghitung nilai rata-rata karena lupa menggunakan tanda kurung. kode yang yang salah:

average = a + b + c / 3

kesalahannya, programemer mengira semua bilangan akan dijumlahkan terlebih dahulu, kemudian dibagi 3. namun, operator pembagian (/) memiliki prioritas lebih tinggi dari pada operator penjumlahan (+), sehingga yang dihitung terlebih dahulu adalah c/3, akibatnya, hasil rata-rata nya menjadi salah.

perbaikannya, menambahkan tanda kurung agar penjumlahan dilakukan terlebih dahulu, kemudian hasilnya di bagi 3, contoh:

average = (a + b + c) / 3

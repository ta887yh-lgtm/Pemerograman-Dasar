```python
# Nama : Tiara Putri Az Zahra
# NIM : 4232501048
# Tanggal : 10, November 2026
# Mata Kuliah : Pemrograman Dasar

# Soal 1 - Daya Generator dan Efisiensi Turbin

# Menghitung daya listrik generator
tegangan = 220
arus = 15

daya = tegangan * arus

print(f"Daya listrik = {daya} W")

# Menghitung efisiensi turbin
energi_input = 500
energi_output = 400

efisiensi = (energi_output / energi_input) * 100

print(f"Efisiensi = {efisiensi:.2f}%")
```

**Cara Kerja Program**

Program dibuat untuk menghitung daya listrik yang dihasilkan oleh generator dan efisiensi energi pada turbin.
1. tegangan = 220 digunakan untuk menyimpan nilai tegangan sebesar 220 Volt, sedangkan arus = 15 digunakan untuk menyimpan arus sebesar 15 Ampere.
2. daya = tegangan * arus digunakan untuk menghitung daya listrik dengan mengalikan tegangan dan arus. Hasilnya adalah 3300 Watt.
3. print(f"Daya listrik = {daya} W") digunakan untuk menampilkan hasil daya listrik menggunakan f-string.
4. energi_input = 500 digunakan untuk menyimpan energi yang diterima turbin, sedangkan energi_output = 400 digunakan untuk menyimpan energi yang dihasilkan turbin.
5. efisiensi = (energi_output / energi_input) * 100 digunakan untuk menghitung efisiensi turbin dengan membagi energi output dengan energi input, kemudian dikalikan 100%. Hasilnya adalah 80%.
6. print(f"Efisiensi = {efisiensi:.2f}%") digunakan untuk menampilkan hasil efisiensi dengan dua angka di belakang koma. Format :.2f berfungsi mengatur jumlah angka desimal menjadi dua.

**Fungsi Setiap Kode**

1. = : Memberikan nilai ke variabel.
2. (*) : Melakukan perkalian.
3. / : Melakukan pembagian.
4. (+) : Melakukan penjumlahan (tidak digunakan dalam program ini).
print() : Menampilkan hasil program.
5. f-string : Menampilkan teks bersama nilai variabel.
6. :.2f : Menampilkan angka dengan dua angka desimal.

```python
# Nama : Tiara Putri Az Zahra
# NIM : 4232501048
# Tanggal : 10, November 2026
# Mata Kuliah : Pemrograman Dasar

# Soal 5 - Katup Pengaman

for i in range(5):
    print(f"\nPengujian ke-{i+1}")

    tekanan_uap = float(input("Masukkan tekanan uap: "))
    status_manual_override = input("Status manual override (True/False): ")

    if tekanan_uap > 150 and status_manual_override == "False":
        print("Katup Terbuka")
    elif status_manual_override == "True":
        print("Katup Mode Manual")
    else:
        print("Katup Tertutup")
```

**Cara kerja Program**

Program tersebut digunakan untuk menentukan kondisi katup pengaman berdasarkan nilai tekanan uap dan status manual override.

1. tekanan_uap = float(input("Masukkan tekanan uap (psi): ")) digunakan untuk meminta pengguna memasukkan nilai tekanan uap. Fungsi input() menerima masukan dari pengguna, sedangkan float() mengubah masukan menjadi tipe data desimal.
2. status_manual_override = False digunakan untuk menentukan status manual override. Nilai False berarti sistem tidak sedang menggunakan mode manual, sedangkan True berarti mode manual sedang aktif.
3. if tekanan_uap > 150 and status_manual_override == False: digunakan untuk memeriksa apakah tekanan uap lebih dari 150 psi dan status manual override bernilai False. Operator logika and memastikan kedua kondisi harus terpenuhi agar katup terbuka.
4. print("Katup Terbuka") digunakan untuk menampilkan status katup terbuka apabila kedua kondisi pada perintah if terpenuhi.
5. elif status_manual_override == True: digunakan untuk memeriksa apakah status manual override bernilai True. Jika kondisi ini terpenuhi, program akan menampilkan status katup mode manual.
6. print("Katup Mode Manual") digunakan untuk menampilkan informasi bahwa katup sedang berada dalam mode manual.
7. else: digunakan untuk menjalankan perintah apabila seluruh kondisi sebelumnya tidak terpenuhi.
8. print("Katup Tertutup") digunakan untuk menampilkan status katup tertutup apabila tekanan uap tidak lebih dari 150 psi dan manual override tidak aktif.

**Fungsi Setiap Kode**
1. input() : Menerima masukan dari pengguna.
2. float() : Mengubah input menjadi bilangan desimal.
3. if : Memeriksa kondisi pertama.
4. elif : Memeriksa kondisi berikutnya.
5. else : Menjalankan perintah jika kondisi sebelumnya tidak terpenuhi.
6. and : Menggabungkan dua kondisi yang harus sama-sama benar.
7. ==	: Membandingkan kesamaan dua nilai.
8. (>) : Membandingkan apakah nilai lebih besar dari nilai lainnya.
9. print() : Menampilkan hasil program.

```python
# Nama : Tiara Putri Az Zahra
# NIM : 4232501048
# Tanggal : 10, November 2026
# Mata Kuliah : Pemrograman Dasar

# Soal 7 - Pemanas Mencapai Target

suhu_saat_ini = 60

while suhu_saat_ini < 85:
    suhu_saat_ini += 5
    print(f"Suhu saat ini: {suhu_saat_ini} C")

    if suhu_saat_ini == 80:
        print("WARNING: Mendekati target!")

print("Target 85 C tercapai.")
```

**Cara Kerja Program**

Program tersebut digunakan untuk menaikkan suhu pada sistem pemanas dari suhu awal 60°C hingga mencapai target 85°C dengan menggunakan perulangan while.
1. suhu_saat_ini = 60 digunakan untuk menentukan dan menyimpan suhu awal sistem pemanas sebesar 60°C ke dalam variabel suhu_saat_ini.
2. while suhu_saat_ini < 85: digunakan untuk membuat perulangan selama suhu saat ini masih kurang dari 85°C. Perulangan akan terus berjalan selama kondisi tersebut bernilai benar.
3. suhu_saat_ini += 5 digunakan untuk menaikkan suhu sebesar 5°C pada setiap perulangan. Operator += berfungsi menambahkan nilai ke variabel yang sudah ada.
4. print(f"Suhu saat ini: {suhu_saat_ini} C") digunakan untuk menampilkan suhu setelah mengalami kenaikan sebesar 5°C. F-string digunakan untuk menampilkan nilai variabel di dalam teks.
5. if suhu_saat_ini == 80: digunakan untuk memeriksa apakah suhu saat ini sudah mencapai tepat 80°C. Operator == digunakan untuk membandingkan kesamaan nilai.
6. print("WARNING: Mendekati target!") digunakan untuk menampilkan pesan peringatan ketika suhu mencapai 80°C karena suhu sudah mendekati target 85°C.
7. print("Target 85 C tercapai.") digunakan untuk menampilkan pesan bahwa suhu telah mencapai target setelah perulangan while selesai.

**Fungsi Setiap Kode**
1. while : Melakukan perulangan selama kondisi bernilai benar.
2. < : Operator perbandingan kurang dari.
3. += : Menambahkan nilai ke variabel.
4. if : Memeriksa suatu kondisi.
5. == : Membandingkan kesamaan nilai.
6.  print() : Menampilkan hasil ke layar.
7. f-string : Menampilkan teks beserta nilai variabel.

```python
# Nama : Tiara Putri Az Zahra
# NIM : 4232501048
# Tanggal : 10, November 2026
# Mata Kuliah : Pemrograman Dasar

# Soal 9 - Pembacaan Vibrasi dengan Sensor Rusak

data = ["1.2", "0.9", "ERR", "3.1", "1.8"]

jumlah_invalid = 0

for i in range(len(data)):
    try:
        nilai = float(data[i])

        if nilai > 2.0:
            print(f"Warning: Unit {i+1} vibrasi tinggi!")

        if nilai > 3.0:
            print("KRITIS: pemeriksaan dihentikan")
            break

    except ValueError:
        print(f"Unit {i+1}: data tidak valid")
        jumlah_invalid += 1
        continue

print(f"Jumlah data tidak valid: {jumlah_invalid}")
```

**Cara Kerja Program**
1. Variabel data menyimpan lima data vibrasi mesin, sedangkan jumlah_invalid menghitung data yang tidak valid.
2. Perulangan for memeriksa setiap data, kemudian float() mengubah data menjadi angka desimal.
3. try-except ValueError menangani data yang tidak dapat diubah menjadi angka. Jika gagal, program menampilkan pesan, menambah penghitung, lalu melanjutkan ke unit berikutnya dengan continue.
4. Jika nilai melebihi 2.0, program menampilkan peringatan. Jika melebihi 3.0, program menampilkan pesan kritis dan menghentikan perulangan dengan break.
5. Setelah perulangan selesai, print() menampilkan jumlah data tidak valid.

**Fungsi Setiap Kode**
1. for : Mengulang pemeriksaan setiap data.
2. range(len(data)) : Menghasilkan nomor indeks data.
3. try : Mencoba menjalankan kode.
4. except ValueError : Menangani kesalahan konversi data.
5. float() : Mengubah teks menjadi angka desimal.
6. continue : Melanjutkan ke data berikutnya.
7. break : Menghentikan perulangan.
8. += 1 : Menambah penghitung sebanyak satu.
9. if : Memeriksa kondisi.
10. print() : Menampilkan hasil.

```python
# Challenge: Bonus Nilai

# Program Pemantauan Suhu pada 5 Unit Boiler

# List untuk menyimpan data suhu boiler
data_suhu = [75.5, 82.0, 90.5, 105.0, 88.5]

# Menghitung jumlah unit dengan suhu tidak normal
jumlah_anomali = 0

# Memeriksa suhu setiap unit menggunakan perulangan for
for i in range(len(data_suhu)):
    try:
        suhu = float(data_suhu[i])

        print(f"\nUnit Boiler {i+1}: {suhu} C")

        # Menentukan kondisi suhu boiler
        if suhu < 80:
            print("Status: Suhu Rendah")
        elif suhu <= 100:
            print("Status: Suhu Normal")
        else:
            print("Status: Suhu Tinggi!")
            jumlah_anomali += 1

    except ValueError:
        print(f"Unit Boiler {i+1}: Data tidak valid")
        jumlah_anomali += 1

# Menampilkan jumlah anomali suhu
print("\nJumlah suhu tidak normal:", jumlah_anomali)
```

**Cara Kerja Program**
1. List data_suhu digunakan untuk menyimpan lima data suhu boiler.
2. Perulangan for memeriksa suhu setiap unit boiler secara berurutan.
3. float() mengubah data menjadi angka desimal.
4. Percabangan if-elif-else menentukan kondisi suhu, yaitu rendah, normal, atau tinggi.
5. Jika suhu melebihi 100°C, program memberikan peringatan dan menambah jumlah anomali.
6. try-except digunakan untuk menangani kesalahan apabila terdapat data yang tidak valid.
7. Setelah pemeriksaan selesai, program menampilkan jumlah suhu yang tidak normal.

**Fungsi Setiap Kode**

1. list / [ ]	Menyimpan data suhu boiler.
2. for	Mengulang pemeriksaan setiap unit.
3. range(len())	Menghasilkan indeks berdasarkan jumlah data.
4. float()	Mengubah data menjadi angka desimal.
5. if-elif-else	Menentukan kondisi suhu boiler.
6. try-except	Menangani kesalahan data.
7. += 1	Menambah jumlah anomali.
8. print()	Menampilkan hasil pemeriksaan.

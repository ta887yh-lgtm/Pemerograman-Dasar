```python
# Nama : Tiara Putri Az Zahra
# NIM : 4232501048
# Tanggal : 10, November 2026
# Mata Kuliah : Pemrograman Dasar

# Soal 1 - Daya Generator dan Efisiensi Turbin

# Menghitung daya listrik generator
tegangan = 220
arus = 15

daya = tegangan * arus

print(f"Daya listrik = {daya} W")

# Menghitung efisiensi turbin
energi_input = 500
energi_output = 400

efisiensi = (energi_output / energi_input) * 100

print(f"Efisiensi = {efisiensi:.2f}%")
```

**Cara Kerja Program**

Program tersebut digunakan untuk menghitung daya listrik dan efisiensi turbin. Pertama, variabel tegangan = 220 digunakan untuk menyimpan nilai tegangan sebesar 220 Volt, sedangkan arus = 15 digunakan untuk menyimpan nilai arus sebesar 15 Ampere. Selanjutnya, program menghitung daya listrik dengan mengalikan nilai tegangan dan arus menggunakan perintah daya = tegangan * arus. Hasil perhitungan tersebut adalah 3300 Watt, kemudian ditampilkan menggunakan perintah print() dengan format f-string.

Berikutnya, variabel energi_input = 500 digunakan untuk menyimpan energi yang diterima turbin, sedangkan energi_output = 400 digunakan untuk menyimpan energi yang dihasilkan turbin. Program kemudian menghitung efisiensi turbin dengan membagi energi output dengan energi input, lalu mengalikannya dengan 100. Hasil perhitungan efisiensi tersebut adalah 80%. Terakhir, perintah print(f"Efisiensi = {efisiensi:.2f}%") digunakan untuk menampilkan hasil efisiensi dengan dua angka di belakang koma. Dengan demikian, program menghasilkan daya listrik sebesar 3300 Watt dan efisiensi turbin sebesar 80,00%.

**Fungsi Setiap Kode**

1. = : Memberikan nilai ke variabel.
2. (*) : Melakukan perkalian.
3. / : Melakukan pembagian.
4. (+) : Melakukan penjumlahan (tidak digunakan dalam program ini).
print() : Menampilkan hasil program.
5. f-string : Menampilkan teks bersama nilai variabel.
6. :.2f : Menampilkan angka dengan dua angka desimal.

```python
# Nama : Tiara Putri Az Zahra
# NIM : 4232501048
# Tanggal : 10, November 2026
# Mata Kuliah : Pemrograman Dasar

# Soal 5 - Katup Pengaman

for i in range(5):
    print(f"\nPengujian ke-{i+1}")

    tekanan_uap = float(input("Masukkan tekanan uap: "))
    status_manual_override = input("Status manual override (True/False): ")

    if tekanan_uap > 150 and status_manual_override == "False":
        print("Katup Terbuka")
    elif status_manual_override == "True":
        print("Katup Mode Manual")
    else:
        print("Katup Tertutup")
```

**Cara kerja Program**

Program tersebut digunakan untuk menentukan kondisi katup pengaman berdasarkan nilai tekanan uap dan status manual override. Pertama, tekanan_uap = float(input("Masukkan tekanan uap (psi): ")) digunakan untuk meminta pengguna memasukkan nilai tekanan uap, kemudian fungsi float() mengubah data masukan menjadi tipe desimal. Selanjutnya, status_manual_override = False digunakan untuk menentukan status manual override, dengan nilai False berarti mode manual tidak aktif dan True berarti mode manual sedang aktif.

Pada bagian percabangan, if tekanan_uap > 150 and status_manual_override == False: digunakan untuk memeriksa apakah tekanan uap lebih dari 150 psi dan mode manual tidak aktif. Operator logika and memastikan bahwa kedua kondisi harus terpenuhi agar program menampilkan tulisan "Katup Terbuka". Selanjutnya, elif status_manual_override == True: digunakan untuk memeriksa apakah mode manual sedang aktif. Jika kondisi tersebut terpenuhi, program akan menampilkan tulisan "Katup Mode Manual". Sementara itu, else: digunakan apabila seluruh kondisi sebelumnya tidak terpenuhi, yaitu ketika tekanan uap tidak lebih dari 150 psi dan mode manual tidak aktif. Dalam kondisi tersebut, program akan menampilkan tulisan "Katup Tertutup".

Dengan demikian, program ini menggunakan percabangan if-elif-else untuk menentukan status katup pengaman berdasarkan tekanan uap dan kondisi manual override.

**Fungsi Setiap Kode**
1. input() : Menerima masukan dari pengguna.
2. float() : Mengubah input menjadi bilangan desimal.
3. if : Memeriksa kondisi pertama.
4. elif : Memeriksa kondisi berikutnya.
5. else : Menjalankan perintah jika kondisi sebelumnya tidak terpenuhi.
6. and : Menggabungkan dua kondisi yang harus sama-sama benar.

```python
# Nama : Tiara Putri Az Zahra
# NIM : 4232501048
# Tanggal : 10, November 2026
# Mata Kuliah : Pemrograman Dasar

# Soal 7 - Pemanas Mencapai Target

suhu_saat_ini = 60

while suhu_saat_ini < 85:
    suhu_saat_ini += 5
    print(f"Suhu saat ini: {suhu_saat_ini} C")

    if suhu_saat_ini == 80:
        print("WARNING: Mendekati target!")

print("Target 85 C tercapai.")
```

**Cara Kerja Program**

Program tersebut digunakan untuk menaikkan suhu pada sistem pemanas dari suhu awal 60°C hingga mencapai target 85°C dengan menggunakan perulangan while.

Program tersebut digunakan untuk menaikkan suhu pada sistem pemanas dari suhu awal 60°C hingga mencapai target 85°C dengan menggunakan perulangan while. Pertama, variabel suhu_saat_ini = 60 digunakan untuk menyimpan suhu awal sistem pemanas sebesar 60°C. Selanjutnya, perintah while suhu_saat_ini < 85 digunakan untuk menjalankan perulangan selama suhu masih kurang dari 85°C.

Pada setiap perulangan, perintah suhu_saat_ini += 5 digunakan untuk menaikkan suhu sebesar 5°C. Setelah suhu dinaikkan, perintah print() digunakan untuk menampilkan suhu saat ini menggunakan f-string. Kemudian, percabangan if suhu_saat_ini == 80 digunakan untuk memeriksa apakah suhu sudah mencapai 80°C. Jika kondisi terpenuhi, program akan menampilkan pesan peringatan "WARNING: Mendekati target!".

Perulangan akan terus berjalan hingga suhu mencapai 85°C. Setelah kondisi while tidak lagi terpenuhi, program akan menampilkan pesan "Target 85 C tercapai." sebagai tanda bahwa suhu telah mencapai target.

**Fungsi Setiap Kode**
1. while : Melakukan perulangan selama kondisi bernilai benar.
2. < : Operator perbandingan kurang dari.
3. += : Menambahkan nilai ke variabel.
4. if : Memeriksa suatu kondisi.
5. == : Membandingkan kesamaan nilai.
6.  print() : Menampilkan hasil ke layar.
7. f-string : Menampilkan teks beserta nilai variabel.

```python
# Nama : Tiara Putri Az Zahra
# NIM : 4232501048
# Tanggal : 10, November 2026
# Mata Kuliah : Pemrograman Dasar

# Soal 9 - Pembacaan Vibrasi dengan Sensor Rusak

data = ["1.2", "0.9", "ERR", "3.1", "1.8"]

jumlah_invalid = 0

for i in range(len(data)):
    try:
        nilai = float(data[i])

        if nilai > 2.0:
            print(f"Warning: Unit {i+1} vibrasi tinggi!")

        if nilai > 3.0:
            print("KRITIS: pemeriksaan dihentikan")
            break

    except ValueError:
        print(f"Unit {i+1}: data tidak valid")
        jumlah_invalid += 1
        continue

print(f"Jumlah data tidak valid: {jumlah_invalid}")
```

**Cara Kerja Program**
1. Variabel data menyimpan lima data vibrasi mesin, sedangkan jumlah_invalid menghitung data yang tidak valid.
2. Perulangan for memeriksa setiap data, kemudian float() mengubah data menjadi angka desimal.
3. try-except ValueError menangani data yang tidak dapat diubah menjadi angka. Jika gagal, program menampilkan pesan, menambah penghitung, lalu melanjutkan ke unit berikutnya dengan continue.
4. Jika nilai melebihi 2.0, program menampilkan peringatan. Jika melebihi 3.0, program menampilkan pesan kritis dan menghentikan perulangan dengan break.
5. Setelah perulangan selesai, print() menampilkan jumlah data tidak valid.

**Fungsi Setiap Kode**
1. for : Mengulang pemeriksaan setiap data.
2. range(len(data)) : Menghasilkan nomor indeks data.
3. try : Mencoba menjalankan kode.
4. except ValueError : Menangani kesalahan konversi data.
5. float() : Mengubah teks menjadi angka desimal.
6. continue : Melanjutkan ke data berikutnya.
7. break : Menghentikan perulangan.
8. += 1 : Menambah penghitung sebanyak satu.
9. if : Memeriksa kondisi.
10. print() : Menampilkan hasil.

```python
# Challenge: Bonus Nilai

# Program Pemantauan Suhu pada 5 Unit Boiler

# List untuk menyimpan data suhu boiler
data_suhu = [75.5, 82.0, 90.5, 105.0, 88.5]

# Menghitung jumlah unit dengan suhu tidak normal
jumlah_anomali = 0

# Memeriksa suhu setiap unit menggunakan perulangan for
for i in range(len(data_suhu)):
    try:
        suhu = float(data_suhu[i])

        print(f"\nUnit Boiler {i+1}: {suhu} C")

        # Menentukan kondisi suhu boiler
        if suhu < 80:
            print("Status: Suhu Rendah")
        elif suhu <= 100:
            print("Status: Suhu Normal")
        else:
            print("Status: Suhu Tinggi!")
            jumlah_anomali += 1

    except ValueError:
        print(f"Unit Boiler {i+1}: Data tidak valid")
        jumlah_anomali += 1

# Menampilkan jumlah anomali suhu
print("\nJumlah suhu tidak normal:", jumlah_anomali)
```

**Cara Kerja Program**
1. List data_suhu digunakan untuk menyimpan lima data suhu boiler.
2. Perulangan for memeriksa suhu setiap unit boiler secara berurutan.
3. float() mengubah data menjadi angka desimal.
4. Percabangan if-elif-else menentukan kondisi suhu, yaitu rendah, normal, atau tinggi.
5. Jika suhu melebihi 100°C, program memberikan peringatan dan menambah jumlah anomali.
6. try-except digunakan untuk menangani kesalahan apabila terdapat data yang tidak valid.
7. Setelah pemeriksaan selesai, program menampilkan jumlah suhu yang tidak normal.

**Fungsi Setiap Kode**

1. list / [ ]	: Menyimpan data suhu boiler.
2. for : Mengulang pemeriksaan setiap unit.
3. range(len()) : Menghasilkan indeks berdasarkan jumlah data.
4. float() : Mengubah data menjadi angka desimal.
5. if-elif-else : Menentukan kondisi suhu boiler.
6. try-except : Menangani kesalahan data.
7. += 1 : Menambah jumlah anomali.
8. print()	: Menampilkan hasil pemeriksaan.
